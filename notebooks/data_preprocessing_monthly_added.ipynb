```python
import pandas as pd
```

## read data and check

```python
nflx = pd.read_csv('NFLX.csv')
print(nflx.head())
```

```python
# transform the date column to datetime format
nflx['Date'] = pd.to_datetime(nflx['Date'])
nflx = nflx.set_index('Date')
print(nflx.head())
```

## get yearly netflix price(use adj close to get average)

```python
# yearly average closing price
nflx_yearly = nflx['Adj Close'].resample('YE').mean().to_frame(name='NFLX_AdjClose_mean')

#extract the year from the index and add it as a column
nflx_yearly['Year'] = nflx_yearly.index.year

# reset the index to make 'Year' a column
nflx_yearly = nflx_yearly.reset_index(drop=True)

nflx_yearly.head()
```

```python
nflx_yearly.head()
nflx_yearly.tail()
nflx_yearly.shape
```

## Gold price year average

```python
xls = pd.ExcelFile("Gold_price_averages_in_a_range_of_currencies_since_1978.xlsx")
xls.sheet_names
```

```python
gold_raw = pd.read_excel(
    "Gold_price_averages_in_a_range_of_currencies_since_1978.xlsx",
    sheet_name="Yearly_Avg",
    header=None
)

gold_raw.head(15)
```

```python
gold_data = gold_raw.iloc[6:, :].copy()

# rename the columns to something more meaningful
gold_data = gold_data.rename(columns={
    2: 'Date',
    3: 'USD'
})

# keep only the date and USD columns
gold_data = gold_data[['Date', 'USD']]
gold_data.head()
```

### Set Date as the index and aggregate by year

```python
# transform the date column to datetime format adn set it as the index
gold_data['Date'] = pd.to_datetime(gold_data['Date'])
gold_data = gold_data.set_index('Date')

# resample the data to yearly frequency and take the mean of the USD column
gold_yearly = gold_data['USD'].resample('YE').mean().to_frame(name='Gold_USD_mean')

# extract the year from the index and add it as a column
gold_yearly['Year'] = gold_yearly.index.year
gold_yearly = gold_yearly.reset_index(drop=True)
gold_yearly.head()
```

## combine with nflx_yearly

```python
combined = pd.merge(
    gold_yearly[['Year', 'Gold_USD_mean']],
    nflx_yearly[['Year', 'NFLX_AdjClose_mean']],
    on='Year',
    how='inner'
)

combined.head()
combined.to_csv("combined_goldandnflx_yearly.csv", index=False)
```

## Monthly NFLX Prices

```python
# Monthly average adj closing price for Netflix
nflx_monthly = nflx['Adj Close'].resample('ME').mean().to_frame(name='NFLX_AdjClose_m_mean')

# Extract year and month
nflx_monthly['Year'] = nflx_monthly.index.year
nflx_monthly['Month'] = nflx_monthly.index.month

# Reset index
nflx_monthly = nflx_monthly.reset_index(drop=True)

nflx_monthly.head()
```

## Monthly gold prices(Still USD)

```python
# Read monthly gold data from the 'Monthly_Avg' sheet
gold_monthly_raw = pd.read_excel(
    "Gold_price_averages_in_a_range_of_currencies_since_1978.xlsx",
    sheet_name="Monthly_Avg",
    header=None
)

gold_monthly_raw.head(15)
```

```python
gold_monthly = gold_monthly_raw.iloc[6:, :].copy()

gold_monthly = gold_monthly.rename(columns={
    2: 'Date',
    3: 'USD'
})

gold_monthly = gold_monthly[['Date', 'USD']]
gold_monthly['Date'] = pd.to_datetime(gold_monthly['Date'])
gold_monthly = gold_monthly.set_index('Date')

# Resample to monthly frequency (month-end), take mean
gold_monthly = gold_monthly['USD'].resample('ME').mean().to_frame(name='Gold_USD_m_mean')

gold_monthly['Year'] = gold_monthly.index.year
gold_monthly['Month'] = gold_monthly.index.month
gold_monthly = gold_monthly.reset_index(drop=True)

gold_monthly.head()
```

## Combine monthly gold and nflx

```python
combined_monthly = pd.merge(
    gold_monthly[['Year', 'Month', 'Gold_USD_m_mean']],
    nflx_monthly[['Year', 'Month', 'NFLX_AdjClose_m_mean']],
    on=['Year', 'Month'],
    how='inner'
)

combined_monthly.head()

combined_monthly.to_csv("combined_goldandnflx_monthly.csv", index=False)
```

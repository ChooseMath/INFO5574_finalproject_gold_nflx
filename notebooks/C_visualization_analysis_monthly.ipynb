```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read the monthly merged dataset
# If this notebook is in a different folder, change the file path accordingly.
df = pd.read_csv("combined_goldandnflx_monthly.csv")

# Create a monthly Date column from Year and Month
# This makes the time-series plot easier to read.
df["Date"] = pd.to_datetime(
    df["Year"].astype(str) + "-" + df["Month"].astype(str).str.zfill(2) + "-01"
)

# Sort data by date
df = df.sort_values("Date")


# 1. Monthly trend plot: Gold price vs Netflix stock price
fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.plot(df["Date"], df["Gold_USD_m_mean"], marker="o", markersize=3, label="Gold Price")
ax1.set_xlabel("Date")
ax1.set_ylabel("Gold Price (USD)")
ax1.tick_params(axis="y")

ax2 = ax1.twinx()
ax2.plot(df["Date"], df["NFLX_AdjClose_m_mean"], marker="s", markersize=3, label="Netflix Stock Price")
ax2.set_ylabel("Netflix Adjusted Close Price")
ax2.tick_params(axis="y")

plt.title("Monthly Trend of Gold Price and Netflix Stock Price")
fig.autofmt_xdate()
fig.tight_layout()
plt.savefig("monthly_trend_gold_vs_nflx.png", dpi=300)
plt.show()


# 2. Scatter plot with regression line
plt.figure(figsize=(8, 6))
sns.regplot(
    x="Gold_USD_m_mean",
    y="NFLX_AdjClose_m_mean",
    data=df
)

plt.xlabel("Gold Price (USD)")
plt.ylabel("Netflix Adjusted Close Price")
plt.title("Relationship Between Monthly Gold Price and Netflix Stock Price")
plt.tight_layout()
plt.savefig("monthly_scatter_regression_gold_vs_nflx.png", dpi=300)
plt.show()


# 3. Correlation heatmap
corr_matrix = df[["Gold_USD_m_mean", "NFLX_AdjClose_m_mean"]].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Monthly Correlation Heatmap")
plt.tight_layout()
plt.savefig("monthly_correlation_heatmap.png", dpi=300)
plt.show()


# Print correlation matrix
print(corr_matrix)
```

The visualization analysis uses the monthly merged dataset. The dataset contains Year, Month, Gold_USD_m_mean, and NFLX_AdjClose_m_mean. A Date column was created from Year and Month so the monthly trend can be plotted clearly.

The monthly trend plot shows that gold prices and Netflix stock prices both increased over the full time period, but they do not always move together month by month. Netflix stock prices show sharper changes, while gold prices move more gradually.

The scatter plot with a regression line shows a positive relationship between monthly gold prices and monthly Netflix adjusted closing prices. However, the points are spread out around the regression line, which means the relationship is not perfectly linear.

The correlation heatmap shows a positive correlation between the two variables. In this monthly dataset, the correlation is about 0.67. This suggests a moderate to strong positive association in the raw monthly data.

Overall, the visualization results suggest that gold prices and Netflix stock prices are positively associated, but the graphs do not prove a causal relationship. The positive relationship may be influenced by long-term time trends or broader market conditions.

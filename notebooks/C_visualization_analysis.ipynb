```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("combined_goldandnflx_yearly.csv")


fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(df["Year"], df["Gold_USD_mean"], marker="o", label="Gold Price")
ax1.set_xlabel("Year")
ax1.set_ylabel("Gold Price (USD)")
ax1.tick_params(axis="y")

ax2 = ax1.twinx()
ax2.plot(df["Year"], df["NFLX_AdjClose_mean"], marker="s", label="Netflix Stock Price")
ax2.set_ylabel("Netflix Adjusted Close Price")
ax2.tick_params(axis="y")

plt.title("Yearly Trend of Gold Price and Netflix Stock Price")
fig.tight_layout()
plt.savefig("trend_gold_vs_nflx.png", dpi=300)
plt.show()


plt.figure(figsize=(8, 6))
sns.regplot(
    x="Gold_USD_mean",
    y="NFLX_AdjClose_mean",
    data=df
)

plt.xlabel("Gold Price (USD)")
plt.ylabel("Netflix Adjusted Close Price")
plt.title("Relationship Between Gold Price and Netflix Stock Price")
plt.tight_layout()
plt.savefig("scatter_regression_gold_vs_nflx.png", dpi=300)
plt.show()


corr_matrix = df[["Gold_USD_mean", "NFLX_AdjClose_mean"]].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=300)
plt.show()


print(corr_matrix)
```

The analysis shows a moderate to strong positive correlation (0.69) between gold prices and Netflix stock prices. This suggests that the two variables generally move in the same direction over time.

However, the scatter plot reveals a wide dispersion of data points, indicating that the relationship is not strictly linear and lacks consistency. Additionally, the time-series trend shows that the two variables do not always move together, with several periods of divergence.

Therefore, while a positive correlation exists, it is likely driven by external factors such as overall market growth and time progression, rather than a direct causal relationship between gold prices and Netflix stock performance.
